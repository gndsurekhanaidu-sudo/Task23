```python
import pandas as pd
```

```python
data = {
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male',
               'Male', 'Female', 'Male', 'Female', 'Male'],

    'Department': ['CSE', 'ECE', 'CSE', 'IT', 'ECE',
                   'CSE', 'IT', 'CSE', 'ECE', 'IT'],

    'City': ['Hyderabad', 'Vizag', 'Hyderabad', 'Chennai',
             'Vizag', 'Hyderabad', 'Chennai', 'Vizag',
             'Hyderabad', 'Chennai'],

    'Grade': ['A', 'B', 'A', 'C', 'B', 'A', 'B', 'A', 'C', 'A'],

    'Status': ['Pass', 'Pass', 'Pass', 'Fail', 'Pass',
               'Pass', 'Fail', 'Pass', 'Pass', 'Pass']
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Gender Frequency:")
print(df['Gender'].value_counts())

print("Department Frequency:")
print(df['Department'].value_counts())

print("City Frequency:")
print(df['City'].value_counts())

print("Grade Frequency:")
print(df['Grade'].value_counts())

print("Status Frequency:")
print(df['Status'].value_counts())
```

```python
print("Gender Proportion:")
print(df['Gender'].value_counts(normalize=True))

print("Department Proportion:")
print(df['Department'].value_counts(normalize=True))

print("City Proportion:")
print(df['City'].value_counts(normalize=True))

print("Grade Proportion:")
print(df['Grade'].value_counts(normalize=True))

print("Status Proportion:")
print(df['Status'].value_counts(normalize=True))
```

```python
for col in df.columns:
    print("\n", col, "Percentage:")
    print(df[col].value_counts(normalize=True) * 100)
```

```python
import matplotlib.pyplot as plt

df['Gender'].value_counts().plot(kind='bar')

plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Frequency')

plt.show()
```
